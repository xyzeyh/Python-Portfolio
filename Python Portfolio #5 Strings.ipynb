**String Type**

```python
# Strings are enclosed in quotation marks

full_name = "Nicole Zeyh C. Flores"
course = "BS Chemical Engineering"

print(full_name)
print(course)

# Checking the data type

print(type(full_name))
```

**Reading Strings**

```python
nickname = input("Enter your nickname: ")

print("Hello,", nickname)
```

______

**Indexing Strings**

```python
course = "BS Chemical Engineering"

print(course[0])
print(course[1])
print(course[2])
```

**note:** 
- *`[0]` = first character*
- *`[1]` = second character*

_____

**Slicing Strings**

```python
course = "BS Chemical Engineering"

print(course[0:2])
print(course[3:11])
```

**note:** *Syntax is `string[start:end]`*

_____

**Looping through Strings**

```python
name = "Nicole"

# Using for loop
for letter in name:
    print(letter)

print()

# Using while loop
index = 0

while index < len(name):

    print(name[index])

    index = index + 1
```

_______

**Concatinating Strings**

```python
first_name = "Nicole"
last_name = "Flores"

full_name = first_name + " " + last_name

print(full_name)
```

**note** *the '+' operator just combines strings together*

______

**String Library**

```python
message = "Congratulations to Nics for Graduating!"

print(message.upper())
print(message.lower())

# Replacing text
print(message.replace("Nics", "Students"))

# Removing spaces
print(message.strip())
```

_____

**String Comparison**

```python
course = "BS Chemical Engineering"

if course == "BS Chemical Engineering":
    print("Correct Course")

# Searching inside strings
if "Engineering" in course:
    print("Engineering found")
```

# Setup #1: Chemical Engineering Student

```python
# =========================================
# STUDENT ACADEMIC ACCOUNT GENERATOR
# =========================================

# -----------------------------------------
# READING STRINGS
# Reading user input
# -----------------------------------------

full_name = input("Enter full name: ")
course = input("Enter course: ")

# -----------------------------------------
# STRIPPING WHITE SPACE
# Removing unnecessary spaces
# -----------------------------------------

full_name = full_name.strip()
course = course.strip()

# -----------------------------------------
# STRING TYPE
# Checking data types
# -----------------------------------------

print("STUDENT INFORMATION")
print("----------------------------")
print("Data Type of Name:", type(full_name))
print("Data Type of Course:", type(course))

# -----------------------------------------
# STRING OPERATIONS
# Using len() to count characters
# -----------------------------------------

print("STRING OPERATIONS")
print("----------------------------")
print("Characters in Name:", len(full_name))
print("Characters in Course:", len(course))

# -----------------------------------------
# INDEXING STRINGS
# Getting initials
# -----------------------------------------

first_initial = full_name[0]
course_initial = course[0]

print("INDEXING STRINGS")
print("----------------------------")
print("First Initial:", first_initial)
print("Course Initial:", course_initial)

# -----------------------------------------
# SLICING STRINGS
# Extracting part of the course name
# -----------------------------------------

course_code = course[0:2]

print("SLICING STRINGS")
print("----------------------------")
print("Course Code:", course_code)

# -----------------------------------------
# CONCATENATING STRINGS
# Creating a welcome message
# -----------------------------------------

welcome_message = "Welcome, " + full_name + "!"

print("CONCATENATING STRINGS")
print("----------------------------")
print(welcome_message)

# -----------------------------------------
# STRING LIBRARY
# Formatting strings
# -----------------------------------------

formatted_name = full_name.title()
formatted_course = course.upper()

print("STRING LIBRARY")
print("----------------------------")
print("Formatted Name:", formatted_name)
print("Formatted Course:", formatted_course)

# -----------------------------------------
# REPLACING TEXT
# Generating username
# -----------------------------------------

username = full_name.lower().replace(" ", ".")

print("REPLACING TEXT")
print("----------------------------")
print("Username:", username)

# -----------------------------------------
# CONCATENATING STRINGS
# Creating academic email
# -----------------------------------------

email = username + "@bicol-u.edu.ph"

print("Academic Email:", email)

# -----------------------------------------
# STRING COMPARISON
# Checking course
# -----------------------------------------

print("STRING COMPARISON")
print("----------------------------")

if course == "BS Chemical Engineering":
    print("Verified Chemical Engineering Student")

else:
    print("Course Verification Failed")

# -----------------------------------------
# SEARCHING IN STRINGS
# Searching for keywords
# -----------------------------------------

print("SEARCHING IN STRINGS")
print("----------------------------")

if "Engineering" in course:
    print("Engineering course detected")

# -----------------------------------------
# LOOPING THROUGH STRINGS
# Counting vowels in the name
# -----------------------------------------

print("LOOPING THROUGH STRINGS")
print("----------------------------")

vowel_count = 0

for letter in full_name.lower():

    if letter == "a":
        vowel_count = vowel_count + 1

    elif letter == "e":
        vowel_count = vowel_count + 1

    elif letter == "i":
        vowel_count = vowel_count + 1

    elif letter == "o":
        vowel_count = vowel_count + 1

    elif letter == "u":
        vowel_count = vowel_count + 1

print("Number of Vowels in Name:", vowel_count)

# -----------------------------------------
# FINAL OUTPUT
# -----------------------------------------

print("FINAL STUDENT ACCOUNT")
print("----------------------------")
print("Student Name:", formatted_name)
print("Course:", formatted_course)
print("Username:", username)
print("Academic Email:", email)
```
