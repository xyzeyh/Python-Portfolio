**Concept of Collection**

**note:** *A list stores multiple values inside one variable.*

```python
subjects = [
    "Plant Design",
    "Reaction Engineering",
    "Process Control",
    "Transport Phenomena"
]

print("Chemical Engineering Subjects")
print("----------------------------")

print(subjects)
```

_______

**List and Definite Loops**

```python
subjects = [
    "Plant Design",
    "Reaction Engineering",
    "Process Control",
    "Transport Phenomena"
]

print("SUBJECT LIST")
print("----------------------------")

for subject in subjects:
    print(subject)
```

_____

**Indexing and Lookups**

```python
subjects = [
    "Plant Design",
    "Reaction Engineering",
    "Process Control",
    "Transport Phenomena"
]

print("First Subject:", subjects[0])
print("Second Subject:", subjects[1])
print("Last Subject:", subjects[3])
```

**note:** *Indexing accesses specific items in a list. The first item starts at index 0.*

______

**List Mutability**

```python
subjects = [
    "Plant Design",
    "Reaction Engineering",
    "Process Control"
]

print("Original List")
print(subjects)

# Modifying an item in the list
subjects[2] = "Chemical Engineering Laboratory"

print("Updated List")
print(subjects)
```

**note:** *Lists are mutable, meaning they can be changed
after creation.*

_______

**List of Functions**

```python
units = [3, 5, 4, 2, 3]

print("Units per Subject")
print("----------------------------")

print(units)

print("Total Subjects:", len(units))
print("Lowest Units:", min(units))
print("Highest Units:", max(units))
print("Total Units:", sum(units))
```

______

**Slicing Lists**

```python
subjects = [
    "Plant Design",
    "Reaction Engineering",
    "Process Control",
    "Transport Phenomena",
    "Research"
]

print("All Subjects")
print(subjects)

print("First Three Subjects")
print(subjects[0:3])

print("Last Two Subjects")
print(subjects[3:5])
```

______

**List Methods**

```python
subjects = [
    "Plant Design",
    "Reaction Engineering",
    "Research"
]

print("Original List")
print(subjects)

# append()

subjects.append("Safety Engineering")

print("After append()")
print(subjects)

# remove()

subjects.remove("Research")

print("After remove()")
print(subjects)
```

______

**Sorting Lists**

```python
subjects = [
    "Transport Phenomena",
    "Plant Design",
    "Research",
    "Reaction Engineering"
]

print("Original List")
print(subjects)

subjects.sort()

print("Sorted List")
print(subjects)
```

______

**Splitting Strings into Lists**

```python
schedule = "Monday Tuesday Thursday Friday"

days = schedule.split()

print(days)
```

______

**Using split() to Parse Strings**

```python
student_record = "2025-01523 Nicole BS-Chemical-Engineering"

record = student_record.split()

print("Student Number:", record[0])
print("Student Name:", record[1])
print("Course:", record[2])
```

**note:** *split() can separate information from a string
into meaningful parts.*

# Student Enrollment, Schedule, and Academic Tracker System

The following application combines all list concepts
into one integrated program.

The system allows the student to:
- manage enrolled subjects
- update subject lists
- analyze subject units
- organize schedules
- sort subjects
- parse student records

This activity demonstrates the practical use
of lists in an academic setting.

```python
# -----------------------------------------
# Student Information
# -----------------------------------------

student_name = "Nicole Zeyh C. Flores"
student_record = "2025-01523 BS-Chemical-Engineering 4th-Year"

# -----------------------------------------
# Parsing Student Record using split()
# -----------------------------------------

record_parts = student_record.split()

student_number = record_parts[0]
course = record_parts[1]
year_level = record_parts[2]

# -----------------------------------------
# Collection of Enrolled Subjects
# -----------------------------------------

subjects = [
    "Plant Design",
    "Reaction Engineering",
    "Process Control",
    "Transport Phenomena"
]

# -----------------------------------------
# Corresponding Subject Units
# -----------------------------------------

units = [3, 5, 4, 3]

# -----------------------------------------
# Displaying Student Information
# -----------------------------------------

print("STUDENT ACADEMIC TRACKER")
print("----------------------------")

print("Student Name:", student_name)
print("Student Number:", student_number)
print("Course:", course)
print("Year Level:", year_level)

# -----------------------------------------
# Lists and Definite Loops
# Displaying Subjects
# -----------------------------------------

print("CURRENT ENROLLED SUBJECTS")
print("----------------------------")

for subject in subjects:
    print(subject)

# -----------------------------------------
# Indexing and Lookup
# -----------------------------------------

print("SUBJECT LOOKUP")
print("----------------------------")

print("First Subject:", subjects[0])
print("Second Subject:", subjects[1])
print("Last Subject:", subjects[-1])

# -----------------------------------------
# List Mutability
# Updating a Subject
# -----------------------------------------

subjects[2] = "Chemical Engineering Laboratory"

print("UPDATED SUBJECT")
print("----------------------------")

print(subjects)

# -----------------------------------------
# append() Method
# Adding New Subjects
# -----------------------------------------

subjects.append("Research")
subjects.append("Safety Engineering")

units.append(2)
units.append(3)

print("SUBJECTS AFTER ADDING")
print("----------------------------")

for subject in subjects:
    print(subject)

# -----------------------------------------
# remove() Method
# Dropping a Subject
# -----------------------------------------

subjects.remove("Reaction Engineering")

print("SUBJECTS AFTER REMOVING")
print("----------------------------")

for subject in subjects:
    print(subject)

# -----------------------------------------
# List Functions
# -----------------------------------------

print("ACADEMIC ANALYSIS")
print("----------------------------")

print("Total Subjects:", len(subjects))
print("Highest Unit Value:", max(units))
print("Lowest Unit Value:", min(units))
print("Total Units:", sum(units))

# -----------------------------------------
# Slicing Lists
# Displaying Major Subjects
# -----------------------------------------

print("MAJOR SUBJECTS")
print("----------------------------")

major_subjects = subjects[0:3]

print(major_subjects)

# -----------------------------------------
# Sorting Lists
# Organizing Subjects Alphabetically
# -----------------------------------------

subjects.sort()

print("SORTED SUBJECT LIST")
print("----------------------------")

for subject in subjects:
    print(subject)

# -----------------------------------------
# Splitting Strings into Lists
# Student Weekly Schedule
# -----------------------------------------

schedule = "Monday Wednesday Friday Saturday"

days = schedule.split()

print("\nCLASS SCHEDULE")
print("----------------------------")

for day in days:
    print(day)

# -----------------------------------------
# FINAL SUMMARY
# -----------------------------------------

print("FINAL ENROLLMENT SUMMARY")
print("----------------------------")

print("Student:", student_name)
print("Enrolled Subjects:", len(subjects))
print("Total Units:", sum(units))
print("Class Days:", days)
```
