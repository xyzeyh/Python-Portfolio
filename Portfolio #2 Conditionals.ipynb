**Fourth Year Subject Grades**

*Conditional Execution #1 : Passed or Failed*

```python
grade = 2.75

if grade <= 3.0:
    print("PASSED")
else:
    print("FAILED")
```

```python
plant_design = 1.25
reaction_engineering = 1.50
process_control = 1.75
transport_phenomena = 1.75
chemical_engineering_laboratory = 1.75
safety_engineering = 1.00
research = 1.50
ave = (plant_design + reaction_engineering + process_control + transport_phenomena + chemical_engineering_laboratory + safety_engineering + research)/7

if (plant_design <= 3.0 and
    reaction_engineering <= 3.0 and
    process_control <= 3.0 and
    transport_phenomena <= 3.0 and
    chemical_engineering_laboratory <= 3.0 and
    safety_engineering <= 3.0 and
    research <= 3.0):
    status = "Eligible for Graduation"

else:
    status = "Not Eligible for Graduation"

print("FOURTH YEAR SUBJECT GRADES")
print("----------------------------")
print(f" Plant Design: {plant_design}")
print(f" Reaction Engineering: {reaction_engineering}")
print(f" Process Control: {process_control}")
print(f" Transport Phenomena: {transport_phenomena}")
print(f" Chemical Engineering Laboratory: {chemical_engineering_laboratory}")
print(f" Safety Engineering: {safety_engineering}")
print(f" Research: {research}")

print("GRADUATION STATUS")
print("----------------------------")
print(f" Average: {ave}")
print(status)
```

**Determining Latin Honors**
