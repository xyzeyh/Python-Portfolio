**While Loop**

```python
count = 1

while count <= 5:

    print("Count:", count)

# Updating the iteration variable
    count = count + 1
```

____

**Infinite Loops**

```python
# =========================================
# USING break which is used to stop the loop
# =========================================

number = 1

while True:

    print("Number:", number)

    number = number + 1

# Stop the loop when number becomes 15
    if number > 15:
        break
```

___

**Continue Statement**

```python
for number in range(1, 6):

# Skip number 3
    if number == 3:
        continue

    print(number)
```

**note:** *the continue statement skips the current iteration and proceeds to the next repitition of the loop*

**Loop Idioms**

**note:** *the example shown nelow used loop idiooms for counting, summing, finding averages, and finding largest or smalles values*

```python
# =========================================
# STUDENT GRADE EVALUATION SYSTEM
# USING ITERATION
# =========================================

# Student Information
full_name = "Nicole Zeyh C. Flores"
course = "BS Chemical Engineering"

# Student Grades
grades = [1.50, 1.75, 1.25, 2.00, 1.50, 1.75, 1.25]

# -----------------------------------------
# Displaying Student Information
# -----------------------------------------

print("STUDENT INFORMATION")
print("----------------------------")
print("Full Name:", full_name)
print("Course:", course)

# -----------------------------------------
# Loop Idiom: Counting
# -----------------------------------------

count = 0

for grade in grades:
    count = count + 1

print("Number of Subjects:", count)

# -----------------------------------------
# Loop Idiom: Summing and Average
# -----------------------------------------

total = 0

for grade in grades:
    total = total + grade

average = total / count

print("Total Grade:", total)
print("Average Grade:", average)

# -----------------------------------------
# Largest and Smallest Values
# -----------------------------------------

highest_grade = None
lowest_grade = None

for grade in grades:

    # Finding highest grade
    # Smaller value means better grade
    if highest_grade is None or grade < highest_grade:
        highest_grade = grade

    # Finding lowest grade
    # Larger value means lower performance
    if lowest_grade is None or grade > lowest_grade:
        lowest_grade = grade

# -----------------------------------------
# Displaying Results
# -----------------------------------------

print("ACADEMIC PERFORMANCE")
print("----------------------------")
print("Highest Grade:", highest_grade)
print("Lowest Grade:", lowest_grade)
```
