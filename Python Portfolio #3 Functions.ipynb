**Student Evaluation System**

```python
# -----------------------------------------
# Student Information
# -----------------------------------------

full_name = "Nicole Zeyh C. Flores"
course = "BS Chemical Engineering"

# -----------------------------------------
# Student Grades
# -----------------------------------------

plant_design = 1.50
reaction_engineering = 1.75
process_control = 1.25
transport_phenomena = 1.50
chemical_engineering_laboratory = 1.75
safety_engineering = 1.25
research = 1.50

# -----------------------------------------
# Fruitful Function
# This function computes the average grade
# Parameters:
#   g1, g2, g3, g4, g5, g6, g7
# Returns:
#   average
# -----------------------------------------

def compute_average(g1, g2, g3, g4, g5, g6, g7):

    average = (
        g1 +
        g2 +
        g3 +
        g4 +
        g5 +
        g6 +
        g7
    ) / 7

    return average


# -----------------------------------------
# Fruitful Function
# This function determines Latin Honors
# Parameter:
#   average
# Returns:
#   honors
# -----------------------------------------

def determine_honors(average):

    if average >= 1.00 and average <= 1.20:
        honors = "Summa Cum Laude"

    elif average >= 1.21 and average <= 1.45:
        honors = "Magna Cum Laude"

    elif average >= 1.46 and average <= 1.75:
        honors = "Cum Laude"

    else:
        honors = "No Latin Honors"

    return honors


# -----------------------------------------
# Void / Non-fruitful Function
# This function only displays information
# It does not return a value
# -----------------------------------------

def display_results(name, course_name, average, honors):

    print("STUDENT INFORMATION")
    print("----------------------------")
    print("Full Name:", name)
    print("Course:", course_name)

    print("\nACADEMIC RESULT")
    print("----------------------------")
    print("Average:", average)
    print("Latin Honors:", honors)


# -----------------------------------------
# Calling the Functions
# Arguments are passed into the functions
# -----------------------------------------

average_grade = compute_average(
    plant_design,
    reaction_engineering,
    process_control,
    transport_phenomena,
    chemical_engineering_laboratory,
    safety_engineering,
    research
)

latin_honors = determine_honors(average_grade)

display_results(
    full_name,
    course,
    average_grade,
    latin_honors
)
```
